```python
import pandas as pd

df = pd.read_csv("data_kantin - data_kantin.csv")

df.head()
```

```python
#Informasi dataset
df.info()

#Statistik deskriptif
df.describe()

#Ukuran dataset
print("Shape:", df.shape)

#Mengecek missing value
print("\nMissing Value:")
print(df.isnull().sum())

#Mengecek duplikat
print("\nJumlah duplikat:", df.duplicated().sum())
```

```python
#Mengubah tanggal menjadi datetime
df["tanggal"] = pd.to_datetime(df["tanggal"])

#Memastikan harga dan terjual bertipe numerik
df["harga"] = pd.to_numeric(df["harga"], errors="coerce")
df["terjual"] = pd.to_numeric(df["terjual"], errors="coerce")
```

```python
print(df.isnull().sum())
```

```python
df["menu"] = df["menu"].fillna("Tidak Diketahui")
```

```python
df["terjual"] = df["terjual"].fillna(df["terjual"].median())
```

```python
df = df.drop_duplicates()

print("Jumlah duplikat setelah cleaning:", df.duplicated().sum())
```

```python
print(df.isnull().sum())
print(df.shape)
```

```python
df_filter = df[df["terjual"] > 20]

df_filter.head()
```

```python
df_sort = df.sort_values(by="terjual", ascending=False)

df_sort.head(10)
```

```python
df["pendapatan"] = df["harga"] * df["terjual"]

df.head()
```

```python
penjualan_menu = df.groupby("menu").agg(
    total_terjual=("terjual", "sum"),
    total_pendapatan=("pendapatan", "sum")
).sort_values("total_terjual", ascending=False)

penjualan_menu
```

```python
penjualan_kategori = df.groupby("kategori").agg(
    total_terjual=("terjual", "sum"),
    total_pendapatan=("pendapatan", "sum")
).sort_values("total_pendapatan", ascending=False)

penjualan_kategori
```

```python
menu_terlaris = penjualan_menu["total_terjual"].idxmax()
jumlah_terjual = penjualan_menu["total_terjual"].max()

print("Menu paling banyak terjual:", menu_terlaris)
print("Jumlah terjual:", jumlah_terjual)
```

```python
kategori_tertinggi = penjualan_kategori["total_pendapatan"].idxmax()
pendapatan_tertinggi = penjualan_kategori["total_pendapatan"].max()

print("Kategori dengan pendapatan terbesar:", kategori_tertinggi)
print("Total pendapatan: Rp", pendapatan_tertinggi)
```

```python
produk_pendapatan_tertinggi = penjualan_menu["total_pendapatan"].idxmax()

print("Produk dengan pendapatan terbesar:",
      produk_pendapatan_tertinggi)
```
